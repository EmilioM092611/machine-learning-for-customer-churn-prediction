## Investigating the `TotalCharges` Feature

The `TotalCharges` feature should contain numerical values. However, it is currently stored as a string, indicating that some values may not be valid numbers.

Before converting its data type, we need to identify the records causing this issue.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Optional configuration for visual improvement

plt.style.use("default")
pd.set_option("display.max_columns", None)
```

```python
df = pd.read_csv("../data/raw/WA_Fn-UseC_-Telco-Customer-Churn.csv")
```

```python
# Display the current data type

display(df["TotalCharges"].dtype)
```

```python
# Identify non-numeric values

invalid_total_charges = df[pd.to_numeric(df["TotalCharges"], errors="coerce").isna()]

display(invalid_total_charges)
```

## Converting `TotalCharges` to Numeric

The `TotalCharges` feature is currently stored as a string due to the presence of empty values.

The feature will be converted to a numerical data type, replacing invalid values with `NaN` for further processing.

```python
# Convert TotalCharges to numeric

df["TotalCharges"] = pd.to_numeric(df["TotalCharges"], errors="coerce")

display(df["TotalCharges"].dtype)
```

## Missing Values After Data Type Conversion

After converting the `TotalCharges` feature, invalid entries become missing values (`NaN`), allowing them to be properly identified and handled.

```python
# Count missing values

display(df.isnull().sum())

# Calculate missing value percentages

missing_values = (
    df.isnull()
      .sum()
      .to_frame(name="Missing Values")
)

missing_values["Percentage"] = (
    missing_values["Missing Values"] / len(df) * 100
).round(2)

display(missing_values)
```

## Handling Missing Values

The `TotalCharges` feature contains 11 missing values that appeared after converting the column to a numerical data type.

These records correspond to customers with a tenure of zero months, indicating that they are new customers who have not yet accumulated any charges.

Since the missing values represent only a very small portion of the dataset, removing these records is considered an appropriate approach.

```python
# Remove rows with missing values

df = df.dropna()

display(df.isnull().sum())
```

## Duplicate Records Validation

After handling missing values, the dataset is inspected again to ensure that no duplicate records are present.

```python
# Verify duplicate records

display(df.duplicated().sum())
```

## Numerical Feature Validation

Numerical variables are inspected to identify potential inconsistencies such as unexpected minimum values or unrealistic ranges.

```python
# Generate descriptive statistics

display(df.describe())
```

## Categorical Feature Validation

Each categorical feature is inspected to verify that its values are consistent and do not contain unexpected categories.

```python
# Display unique values for categorical features

for column in df.select_dtypes(include=["object", "string"]).columns:
    print(f"\n{column}")
    print(df[column].unique())
```

## Final Dataset Validation

The dataset is validated after the cleaning process to ensure that all variables have the correct data types and that no missing values remain.

```python
# Display dataset information

display(df.info())

# Display dataset shape

display(df.shape)
```
