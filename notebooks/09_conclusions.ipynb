# Conclusions

## Objective

This phase summarizes the technical and business findings obtained throughout the customer churn prediction project.

The conclusions focus on model performance, customer churn patterns, business implications, limitations, and potential future improvements.

## Project Summary

This project developed a supervised machine learning solution to predict customer churn using the IBM Telco Customer Churn dataset.

The project followed a complete data science workflow:

1. Business understanding
2. Data understanding
3. Data cleaning
4. Exploratory data analysis
5. Feature engineering
6. Model training
7. Model evaluation
8. Model interpretation
9. Business conclusions

## Technical Conclusions

Three classification algorithms were evaluated:

- Logistic Regression
- Decision Tree
- Random Forest

Logistic Regression achieved the strongest overall performance after hyperparameter tuning.

The final Logistic Regression model achieved:

- Accuracy: 79.67%
- Precision: 63.17%
- Recall: 56.42%
- F1-Score: 59.60%
- ROC-AUC: 83.49%

During cross-validation, the tuned Logistic Regression model achieved a ROC-AUC of 84.62%.

These results indicate that the model has a reasonable ability to distinguish between customers who are likely to churn and customers who are likely to remain.

## Customer Churn Insights

Model interpretation identified several variables strongly associated with customer churn.

The most relevant factors included:

- Customer tenure
- Contract type
- Internet service type
- Monthly charges
- Total charges
- Payment method
- Streaming services

Shorter customer tenure and month-to-month contracts were strongly associated with higher churn risk.

Longer tenure and two-year contracts were associated with lower churn risk.

The permutation importance analysis identified tenure as the most influential feature in the model.

## Business Recommendations

Based on the analysis, the company could consider the following actions:

### 1. Prioritize new customers

Customers with short tenure represent an important retention opportunity.

The company could implement targeted onboarding and early-retention campaigns during the first months of the customer relationship.

### 2. Encourage longer-term contracts

Month-to-month customers showed a stronger association with churn, while two-year contracts were associated with lower churn.

The company could consider incentives for customers who transition to longer-term contracts.

### 3. Identify high-risk customer profiles

The model can be used to prioritize customers with characteristics associated with higher churn probability.

Retention resources could therefore be focused on customers with the highest predicted risk.

### 4. Investigate service-related churn

Internet service type and several additional services appeared among the relevant predictors.

The company could investigate whether service quality, pricing, or customer expectations contribute to churn in these segments.

### 5. Use predictive modeling for proactive retention

Instead of reacting after customers leave, the company could use churn predictions to identify customers who may require intervention before cancellation occurs.

## Limitations

This project has several limitations.

First, the dataset represents a specific customer population and may not generalize to other companies or industries.

Second, the model identifies statistical associations rather than causal relationships.

Third, the dataset does not include temporal information that would allow customer behavior to be analyzed over time.

Fourth, the model performance may be improved by testing additional algorithms, feature engineering strategies, and hyperparameter configurations.

Finally, the current model does not incorporate the financial cost of false positives and false negatives.

## Future Improvements

Several improvements could be explored in future iterations:

- Test additional classification algorithms such as Gradient Boosting, XGBoost, or LightGBM.
- Perform more extensive hyperparameter optimization.
- Apply threshold optimization based on business costs.
- Use SHAP for more advanced model interpretability.
- Perform cross-validation with additional evaluation metrics.
- Analyze customer lifetime value together with churn probability.
- Incorporate historical customer behavior and temporal features.
- Deploy the model as an API or business application.
- Monitor model performance after deployment.

## Business Questions

The following questions were defined at the beginning of the project and are answered below using the exploratory analysis and model interpretation results.

### 1. Which customer characteristics are most strongly associated with churn?

The analysis identified customer tenure, contract type, internet service type, monthly charges, total charges, and payment method as relevant characteristics associated with customer churn.

Among these variables, tenure was particularly important according to the permutation importance analysis.

Customers with shorter tenure and month-to-month contracts showed a stronger association with churn.

### 2. Which characteristics are associated with customer retention?

Longer customer tenure and longer-term contracts were associated with lower churn risk.

In particular, customers with two-year contracts showed a negative association with churn in the Logistic Regression model.

These characteristics suggest that established customer relationships and longer contractual commitments are associated with greater customer retention.

### 3. Which customer profiles should the company prioritize for retention campaigns?

The company should prioritize customers with a combination of characteristics associated with higher predicted churn probability.

In particular, retention efforts could focus on newer customers, customers with month-to-month contracts, and customers whose service and billing characteristics are associated with higher churn risk.

The model can be used to rank customers according to predicted churn probability, allowing retention resources to be allocated more efficiently.

### 4. Are contract type and tenure important predictors of churn?

Yes.

Tenure was the most influential feature according to permutation importance.

Contract type was also among the most relevant variables, with month-to-month contracts showing a positive association with churn and two-year contracts showing a negative association.

These findings indicate that both customer relationship duration and contract structure are important predictive factors.

### 5. Which services or customer characteristics appear to be associated with higher churn risk?

Fiber optic internet service and several service-related variables showed associations with higher predicted churn.

Electronic check payment method and certain streaming and support service configurations also appeared among the relevant predictors.

However, these relationships represent statistical associations identified by the model and should not be interpreted as evidence that these services directly cause churn.

### 6. How can these findings support customer retention strategies?

The findings can support a proactive retention strategy by allowing the company to identify customers who are more likely to churn before they leave.

The company could use the model to prioritize retention campaigns, focus on newer customers, encourage longer-term contracts, and investigate service or pricing factors associated with higher churn risk.

The model should be used as a decision-support and prioritization tool rather than as an automatic decision-making system.

## Final Business Statement

The analysis demonstrates that machine learning can support proactive customer retention by identifying customers with a higher probability of churn.

The final model achieved a ROC-AUC of 83.49% on the test set and provided interpretable insights into the characteristics associated with customer churn.

The results suggest that customer tenure and contract type are particularly relevant factors for identifying churn risk.

From a business perspective, the model could be used as a prioritization tool for retention campaigns, allowing the company to focus its resources on customers who are more likely to leave.
